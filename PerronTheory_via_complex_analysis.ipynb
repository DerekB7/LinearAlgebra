**lemma 1:**  
If $A\geq B\geq \mathbf 0$, then $\rho(A) \geq \rho(B)$  
*remark:* this is the best be we can do with these conditions.  However if $A$ is positive and $A\neq B$ then we can ultimately can sharpen this to $\rho(A) \gt \rho(B)$ which is shown at the end.   

$A\geq B \geq \mathbf 0$ somewhat trivially implies $A^2\geq AB \geq B^2 \geq \mathbf 0$ by checking dot products (summing over a point-wise bound) and by induction $A^k \geq B^k$ 
$\implies \text{trace}\big( A^k\big) \geq \text{trace}\big(B^k\big)\geq 0$ for all $k\in \mathbb N$ (and this is an equality of $n=n$ in the special case of $k=0$).     

Let $A,B$ have characteristic polynomials $p_A$ and $p_B$ and spectral radii of $R_A, R_B$ respectively.  Then in the annulus $\big\{z: R_A \lt \vert z\vert\lt \infty\big\}$ the rational function $r_A(z) = \frac{p_A'(z)}{p_A(z)}$, has Laurent Series   
$ = \sum_{k=1}^\infty \text{trace}\big(A^{k-1}\big)\cdot z^{-k}\text{     }$  (check Cauchy's Integral Formula and the fact that $r_A\big(\infty\big)=0$)  
which implies we have analytic $f_A: \big(0, R_A^{-1}\big)\longrightarrow \mathbb C$ given by  
$f_A(z) = r_A\big(z^{-1}\big)= \sum_{k=1}^\infty \text{trace}\big(A^{k-1}\big)\cdot z^{k}$, where $0$ is a removable singularity that we've defined over.  I.e. $f_A$ converges absolutely for $\vert z \vert \lt R_A^{-1}$ and has a pole at some $\vert z \vert = R_A^{-1}$. and for $z\in \big(0, R_A^{-1}\big)$ we have  

$\big \vert \sum_{k=1}^\infty \text{trace}\big(B^{k-1}\big)\cdot z^{k}\big \vert \leq \sum_{k=1}^\infty \text{trace}\big(B^{k-1}\big)\cdot \big \vert z \big \vert^{k}\leq \sum_{k=1}^\infty \text{trace}\big(A^{k-1}\big)\cdot \big \vert z \big \vert^{k}\lt \infty$  
so $f_B$ converges absolutely in $\big(0,R_A^{-1}\big)\implies R_B^{-1}\geq R_A^{-1}\implies R_B\leq R_A$  

**extension:**  
if $A\gt \mathbf 0$ then its spectral radius $\rho(A)=R_A$ is given by given by a positive eigenvalue (the Perron root) and any non-positive eigenvalues have modulus $\lt \rho(A)$.  As above we have   

$f_A: \big(0, R_A^{-1}\big)\longrightarrow \mathbb C$ given by  
$f_A(z) = r_A\big(z^{-1}\big)= \sum_{k=1}^\infty \text{trace}\big(A^{k-1}\big)\cdot z^{k}$
and for some $\vert w \vert =R_A^{-1}$ the underlying idea is 
$\infty = \big \vert f_A(w) \big \vert\leq f_A\Big(\big \vert w\big \vert \Big)$ by triangle inequality (applied to power series coefficients) 
$\implies w:= R_A^{-1}$ is a pole, i.e. $\rho(A)$ is an eigenvalue of any positive matrix $A$ with spectral radius $\rho(A)$.  However this approach runs into technical nits about boundary behavior of an analytic function on its disc of convergence; a better approach sidesteps these issues.  Suppose $r_A\big(z^{-1}\big)$ has a pole at some $\omega\in \partial B\big(0, R_A^{-1}\big)$, the unit circle. Then for any $z\in B\big(0, R_A^{-1}\big)\cap B\big(\omega,\delta\big)$ we have $\big \vert f_A(z)\big \vert \gt K$ for any $K\gt 0$ by choosing $\delta \gt 0$ small enough. Via triangle inequality, this implies $\big\vert f_A(z)\big\vert \gt K$ hence $\big\vert r_A\big(z^{-1}\big)\big\vert \gt K$ for $z\in B\big(0, R_A^{-1}\big)\cap B\big(R_A^{-1},\delta\big)\cap \mathbb R\implies R_A^{-1}$ is a pole.  $\big[$ i.e. $r_A\big(z^{-1}\big):\mathbb C_\infty\longrightarrow \mathbb C_\infty$ is continuous hence if $R_A^{-1}$ is not a pole, then $r_A\big(z^{-1}\big)$ has a bounded modulus in a neighborhood of $R_A^{-1}\big]$.  
 


If such a matrix $A$ (also) has eigenvalue $e^{i\theta}\rho(A)$ for $\theta \in \big(0, 2\pi\big)$ then for any $\delta \gt 0$ small enough  
$A\mapsto \frac{1}{1-\delta}\big(A-\rho(A)\cdot\delta\cdot I\big)$ is another positive matrix.  This map sends an eigenvalue in the (not open) right-slit plane $\mathbb C-\mathbb R_{\gt 0} \to \mathbb C-\mathbb R_{\gt 0}$ [in particular it sends $\leq 0$ eigenvalues to $\leq 0$ eigenvalues and sends non-real to non-real].  Now it preserves positive eigenvalues and in particular $\rho(A) \mapsto \frac{1}{1-\delta}\big(\rho(A)\cdot (1-\delta)\big)= \rho(A)$ but   
$e^{i\theta}\rho(A)\mapsto \frac{1}{1-\delta}\big(\rho(A)\cdot (e^{i\theta}-\delta)\big)$ which has modulus $\frac{1}{1-\delta}\rho(A) \cdot \big \vert (e^{i\theta}-\delta)\big\vert \gt \frac{1}{1-\delta}\rho(A)\cdot \big \vert 1-\delta\big\vert=\rho(A)$ by (reverse) triangle inequality.  Hence $\frac{1}{1-\delta}\big(A-\rho(A)\cdot\delta \cdot I\big)$ is a positive matrix with spectral radius $\gt $ its largest positive eigenvalue, a contradiction.  



**Main Perron Theory**  $\rho(A)$ **is a simple eigenvalue for** $A\gt \mathbf 0$ and $\ker \big(\rho(A)I-A\big)$ has a positive vector in it   
Proceed by induction on $n$.  The Base Case for $n=1$ trivially holds.  

Inductive Case:  
To simplify we may divide out $\rho(A)$ and WLOG assume $\rho(A)=1$.  

**(a)** The Perron root is simple  
If we consider any $n-1\times n-1$ principal submatrix and zero out all other components of $A$ and call the resulting matrix $B_i$ (i.e. zeroing out all components in row and column $i$), then $B_i\leq A\implies \rho(B_i)\leq 1=\rho(A)$ by lemma 1.  
To show $\rho(A)=1$ is a simple eigenvalue we may look at the characteristic polynomial of $\big (I-A\big)$ and check that the algebraic multiplicity of eigenvalue $0$ is one-- which is equivalent to showing that the sum over all $n-1\times n-1$ principal minors of $\big(I-A\big)$ is non-zero, i.e. $\sum_{i=1}^n \det\big(I_{n-1}-B_i\big)\neq 0$.  Per the preceding paragraph we have $\det\big(I_{n-1}-B_i\big)\geq 0$ since $\big(I_{n-1}-B_i\big)$ has all of its eigenvalues in the closed right half-plane (so the product of real roots is $\geq 0$ and non-real roots come in conjugate pairs hence their product is $\gt 0$).  Thus if we show that (at least one of) these determinants is non-zero, then the sum over principal minors is positive and the Perron root for $A$ is simple and we prove this as follows.  

$\det\big(I-A\big)=0\implies \det\big(I_{n-1}-B_n\big)\neq 0$  
suppose not.  By Lemma 1, we know $1$ is the Perron root of $B_{n}$ which is simple by induction hypothesis applied to the leading $n-1\times n-1$ principle submatrix of $B_n$, and that submatrix's transpose (i.e. leading $n-1\times n-1$ principle submatrix of $B_{n}^T$) has an associated positive (Perron) eigenvector $\mathbf v$ by induction hypothesis; thus $\begin{bmatrix}\mathbf v\\ 0\end{bmatrix}^T A= \begin{bmatrix}\mathbf v\\ \eta\end{bmatrix}^T$ for some $\eta \gt 0$ by positivity of $A$.  Also $A$ has an eigenvector $A\mathbf x =\mathbf x=\begin{bmatrix}\mathbf x_{n-1}\\ x_n\end{bmatrix}$ where at least one component is (chosen to be) positive.  Now compute the same thing two different ways 

$\begin{bmatrix}\mathbf v\\ \eta\end{bmatrix}^T\mathbf x=\left(\begin{bmatrix}\mathbf v\\ 0\end{bmatrix}^T A\right) \mathbf x=\begin{bmatrix}\mathbf v\\ 0\end{bmatrix}^T A \mathbf x= \begin{bmatrix}\mathbf v\\ 0\end{bmatrix}^T\big( A \mathbf x\big)=\begin{bmatrix}\mathbf v\\ 0\end{bmatrix}^T \mathbf x $   

$\implies \eta\cdot x_n=0\implies x_n=0\implies B_{n}\begin{bmatrix}\mathbf x_{n-1}\\ 0\end{bmatrix}=\begin{bmatrix}\mathbf x_{n-1}\\ 0\end{bmatrix}\implies \mathbf x_{n-1} \gt \mathbf 0$ where $\mathbf x_{n-1}$ is the Perron vector for the leading $n-1\times n-1$ principle submatrix of $A$ so must be positive by induction hypothesis but that implies $\begin{bmatrix}\mathbf x_{n-1}\\0 \end{bmatrix}=A\begin{bmatrix}\mathbf x_{n-1}\\0 \end{bmatrix} \gt \mathbf 0$ by positivity of $A$ and non-negativity of $\mathbf x$ which is also impossible.  

Conclude $\sum_{i=1}^n \det\big(I_{n-1}-B_i\big)\gt 0$ and the Perron root of $A$ is simple.    


**(b)** $A\gt \mathbf 0$ **has a positive eigenvector**    
by simplicity of eigenvalue 1  
$\mathbb R^n = \ker \big(I-A\big)\oplus \text{image }\big(I-A\big)$  
$A\mathbf B = \mathbf B \begin{bmatrix} 1 & \mathbf 0^T\\ \mathbf 0 & A_{n-1}\end{bmatrix}$ 
with $\rho\big(A_{n-1}\big)\lt 1$ so  $A_{n-1}^k\to \mathbf 0$ ref e.g.   
https://math.stackexchange.com/questions/4303957/schur-stability-without-the-jordan-canonical-form/ though it is also implied by Cayley-Hamilton, and of course the Gelfand formula implies it as well (ref top of 'MacCluer_chp5.ipynb')    

$\implies A^{\infty}=\lim_{k\to\infty} A^k \geq 0 \implies A^{\infty} =  \mathbf {xy}^T$  
since the limit has rank one, where we may choose $\mathbf x,\mathbf y\geq \mathbf 0$ (if not the limiting matrix is not non-negative) and $\text{trace}\big(\mathbf {xy}^T\big)=1$.  $A$ and $\mathbf {xy}^T$ commute $\big[$ e.g. somewhat overkill $T:M_n(\mathbb R) \rightarrow M_n(\mathbb R)$ given by $T(X) =AX-XA$  is linear hence continuous and $T\big(A^k\big)=\mathbf 0$ for all $k \implies \mathbf 0= \lim_{k\to \infty}T\big(A^k\big)=T\big(\lim_{k\to \infty}A^k\big)=T\big(\mathbf {xy}^T\big)\big]$ so ($1$-dimensional) $\text{image }\mathbf {xy}^T$ is $A$-invariant $\implies \mathbf 0\lt A\mathbf x = \lambda \mathbf x \implies \lambda =1$ and $\mathbf x$ is a positive vector by positivity of $A$ (and the fact that $\mathbf {xy}^T$ is positive implies $\mathbf y$ is positive; alternatively re-run the same logic with $A^T$ to deduce $\mathbf y$ is a positive vector).  


**sharpening of lemma 1:**  
If $A\geq B\geq \mathbf 0$, where $A \neq B$ and $A\gt \mathbf 0$ then $\rho(A) \gt \rho(B)$  
$C :=\frac{A+B}{2}$ is a positive matrix and $\rho(C) \geq \rho(B)$ by lemma $1$.  It remains to prove $\rho(A)\gt \rho(C)$ and we may do so with a maximum principle type argument.     

As before we may divide out the Perron root of $A$ and reduce this to proving $\rho(C)\lt 1= \rho(A)$.
$D:= \text{diag}(\mathbf v)$ where $\mathbf v$ is the (positive) Perron vector of $A$.  
$\implies D^{-1}CD\leq D^{-1}AD$ where $D^{-1}AD$ is stochastic and $D^{-1}CD$ is a positive matrix where each row sums to $\leq 1$ and this is strict in at least one row.   
By Perron Theory $D^{-1}CD$ has a positive Perron vector $\lambda \mathbf x =(D^{-1}CD)\mathbf x$. Now $\lambda =1\implies \mathbf x\propto \mathbf 1$ because the maximum component of $\mathbf x$ is written as a (sub)convex combination of components of $\mathbf x$ with all weights positive, i.e.  
$\text{maximum }x_i = \sum_{k=1}^n w_k\cdot x_k\leq \sum_{k=1}^n w_k\cdot x_i \leq x_i\text{ with equality }\implies x_1= x_2=\dots =x_n \text{  since all }w_k \gt 0$   
but $D^{-1}CD\mathbf 1 \neq \mathbf 1$ since one of its rows sums to $\lt 1$.  Conclude $\rho(C)\lt 1 = \rho(A)$.
